```python
# ============================================================
# LAB 12 — Tool-grounded Single-Agent System
# Student Lab Checker
# ============================================================

import json
from datetime import datetime

# ============================================================
# TEST CASES
# ============================================================

test_cases = [

    # 1. Повністю правильний кейс
    {
        "id": "case_1",
        "repo_url": "https://github.com/user/lab12",
        "release_tag": "lab-12",
        "files": [
            "README.md",
            "requirements.txt",
            "agent.py",
            "tools.py",
            "tool_logger.py"
        ]
    },

    # 2. Відсутній repo URL
    {
        "id": "case_2",
        "repo_url": "",
        "release_tag": "lab-12",
        "files": [
            "README.md"
        ]
    },

    # 3. Немає release tag
    {
        "id": "case_3",
        "repo_url": "https://github.com/user/lab12",
        "release_tag": "",
        "files": [
            "README.md",
            "requirements.txt"
        ]
    },

    # 4. Неправильний release tag
    {
        "id": "case_4",
        "repo_url": "https://github.com/user/lab12",
        "release_tag": "wrong-tag",
        "files": [
            "README.md",
            "requirements.txt"
        ]
    },

    # 5. Немає файлів
    {
        "id": "case_5",
        "repo_url": "https://github.com/user/lab12",
        "release_tag": "lab-12",
        "files": []
    },

    # 6. Відсутній logger
    {
        "id": "case_6",
        "repo_url": "https://github.com/user/project",
        "release_tag": "lab-12",
        "files": [
            "README.md",
            "requirements.txt",
            "agent.py",
            "tools.py"
        ]
    },

    # 7. Лише README
    {
        "id": "case_7",
        "repo_url": "https://github.com/user/project",
        "release_tag": "lab-12",
        "files": [
            "README.md"
        ]
    },

    # 8. Немає requirements
    {
        "id": "case_8",
        "repo_url": "https://github.com/user/project",
        "release_tag": "lab-12",
        "files": [
            "README.md",
            "agent.py",
            "tools.py",
            "tool_logger.py"
        ]
    },

    # 9. Немає tools.py
    {
        "id": "case_9",
        "repo_url": "https://github.com/user/project",
        "release_tag": "lab-12",
        "files": [
            "README.md",
            "requirements.txt",
            "agent.py",
            "tool_logger.py"
        ]
    },

    # 10. Немає agent.py
    {
        "id": "case_10",
        "repo_url": "https://github.com/user/project",
        "release_tag": "lab-12",
        "files": [
            "README.md",
            "requirements.txt",
            "tools.py",
            "tool_logger.py"
        ]
    },

    # 11. noisy repo URL
    {
        "id": "case_11",
        "repo_url": " github.com/user/lab12 ",
        "release_tag": "lab-12",
        "files": [
            "README.md",
            "requirements.txt",
            "agent.py",
            "tools.py",
            "tool_logger.py"
        ]
    },

    # 12. Uppercase tag
    {
        "id": "case_12",
        "repo_url": "https://github.com/user/lab12",
        "release_tag": "LAB-12",
        "files": [
            "README.md",
            "requirements.txt",
            "agent.py",
            "tools.py",
            "tool_logger.py"
        ]
    },

    # 13. Дублікати файлів
    {
        "id": "case_13",
        "repo_url": "https://github.com/user/lab12",
        "release_tag": "lab-12",
        "files": [
            "README.md",
            "README.md",
            "requirements.txt",
            "agent.py",
            "tools.py",
            "tool_logger.py"
        ]
    },

    # 14. Невідомі додаткові файли
    {
        "id": "case_14",
        "repo_url": "https://github.com/user/lab12",
        "release_tag": "lab-12",
        "files": [
            "README.md",
            "requirements.txt",
            "agent.py",
            "tools.py",
            "tool_logger.py",
            "notes.txt",
            "random.py"
        ]
    },

    # 15. Порожній список файлів + неправильний tag
    {
        "id": "case_15",
        "repo_url": "https://github.com/user/lab12",
        "release_tag": "test",
        "files": []
    },

    # 16. Взагалі порожній input
    {
        "id": "case_16",
        "repo_url": "",
        "release_tag": "",
        "files": []
    },

    # 17. Немає README
    {
        "id": "case_17",
        "repo_url": "https://github.com/user/project",
        "release_tag": "lab-12",
        "files": [
            "requirements.txt",
            "agent.py",
            "tools.py",
            "tool_logger.py"
        ]
    },

    # 18. Дуже noisy files
    {
        "id": "case_18",
        "repo_url": "https://github.com/user/project",
        "release_tag": "lab-12",
        "files": [
            " README.md ",
            "TOOLS.PY",
            "agent.PY",
            "tool_logger.py "
        ]
    },

    # 19. Частково правильний кейс
    {
        "id": "case_19",
        "repo_url": "https://github.com/user/project",
        "release_tag": "lab-12",
        "files": [
            "README.md",
            "requirements.txt"
        ]
    },

    # 20. Ще один повністю правильний кейс
    {
        "id": "case_20",
        "repo_url": "https://github.com/user/final-project",
        "release_tag": "lab-12",
        "files": [
            "README.md",
            "requirements.txt",
            "agent.py",
            "tools.py",
            "tool_logger.py"
        ]
    }

]

# ============================================================
# LOGGER
# ============================================================

tool_logs = []

def log_tool_call(
    task_id,
    tool_name,
    tool_input,
    tool_output,
    success=True,
    error=None
):

    log_entry = {
        "timestamp": datetime.now().isoformat(),
        "task_id": task_id,
        "tool_name": tool_name,
        "input": tool_input,
        "output": tool_output,
        "success": success,
        "error": error
    }

    tool_logs.append(log_entry)

# ============================================================
# TOOLS
# ============================================================

def check_required_links(submission):

    repo_exists = bool(submission.get("repo_url"))

    return {
        "repo_exists": repo_exists
    }


def validate_release_tag(tag):

    valid = tag == "lab-12"

    return {
        "valid_tag": valid
    }


def check_repo_structure(file_list):

    required_files = [
        "README.md",
        "requirements.txt",
        "agent.py",
        "tools.py",
        "tool_logger.py"
    ]

    missing_files = []

    for file in required_files:
        if file not in file_list:
            missing_files.append(file)

    return {
        "missing_files": missing_files,
        "structure_valid": len(missing_files) == 0
    }

# ============================================================
# BASELINE (WITHOUT TOOLS)
# ============================================================

def baseline_llm(submission):

    if submission["repo_url"] and submission["release_tag"]:
        return "Submission looks correct"

    return "Submission may have problems"

# ============================================================
# SINGLE AGENT WITH TOOLS
# ============================================================

def agent(submission):

    task_id = submission["id"]

    final_result = {
        "repo_check": None,
        "tag_check": None,
        "structure_check": None,
        "status": None
    }

    # ========================================================
    # TOOL 1 — LINKS
    # ========================================================

    try:

        repo_result = check_required_links(submission)

        log_tool_call(
            task_id=task_id,
            tool_name="check_required_links",
            tool_input=submission,
            tool_output=repo_result,
            success=True
        )

        final_result["repo_check"] = repo_result

    except Exception as e:

        log_tool_call(
            task_id=task_id,
            tool_name="check_required_links",
            tool_input=submission,
            tool_output=None,
            success=False,
            error=str(e)
        )

    # ========================================================
    # TOOL 2 — RELEASE TAG
    # ========================================================

    try:

        tag_result = validate_release_tag(
            submission["release_tag"]
        )

        log_tool_call(
            task_id=task_id,
            tool_name="validate_release_tag",
            tool_input=submission["release_tag"],
            tool_output=tag_result,
            success=True
        )

        final_result["tag_check"] = tag_result

    except Exception as e:

        log_tool_call(
            task_id=task_id,
            tool_name="validate_release_tag",
            tool_input=submission["release_tag"],
            tool_output=None,
            success=False,
            error=str(e)
        )

    # ========================================================
    # TOOL 3 — STRUCTURE
    # ========================================================

    try:

        structure_result = check_repo_structure(
            submission["files"]
        )

        log_tool_call(
            task_id=task_id,
            tool_name="check_repo_structure",
            tool_input=submission["files"],
            tool_output=structure_result,
            success=True
        )

        final_result["structure_check"] = structure_result

    except Exception as e:

        log_tool_call(
            task_id=task_id,
            tool_name="check_repo_structure",
            tool_input=submission["files"],
            tool_output=None,
            success=False,
            error=str(e)
        )

    # ========================================================
    # FINAL DECISION
    # ========================================================

    repo_ok = final_result["repo_check"]["repo_exists"]
    tag_ok = final_result["tag_check"]["valid_tag"]
    structure_ok = final_result["structure_check"]["structure_valid"]

    if repo_ok and tag_ok and structure_ok:
        final_result["status"] = "correct"

    elif repo_ok or tag_ok or structure_ok:
        final_result["status"] = "partly_correct"

    else:
        final_result["status"] = "wrong"

    return final_result

# ============================================================
# RUN TEST CASES
# ============================================================

baseline_outputs = []
agent_outputs = []

print("=" * 60)
print("BASELINE RESULTS")
print("=" * 60)

for case in test_cases:

    result = baseline_llm(case)

    baseline_outputs.append(result)

    print(case["id"], "->", result)

print("\n")

print("=" * 60)
print("AGENT RESULTS")
print("=" * 60)

for case in test_cases:

    result = agent(case)

    agent_outputs.append(result)

    print(case["id"])
    print(json.dumps(result, indent=2, ensure_ascii=False))
    print("-" * 40)

# ============================================================
# METRICS
# ============================================================

successful_calls = sum(
    1 for log in tool_logs
    if log["success"]
)

total_calls = len(tool_logs)

tool_call_success_rate = successful_calls / total_calls

average_tool_calls = total_calls / len(test_cases)

correct_answers = sum(
    1 for result in agent_outputs
    if result["status"] == "correct"
)

print("\n")
print("=" * 60)
print("METRICS")
print("=" * 60)

print("Tool call success rate:", tool_call_success_rate)
print("Average tool calls per task:", average_tool_calls)
print("Correct answers:", correct_answers)

# ============================================================
# SAVE LOGS
# ============================================================

with open("tool_logs_lab12.jsonl", "w", encoding="utf-8") as f:

    for log in tool_logs:

        f.write(
            json.dumps(log, ensure_ascii=False) + "\n"
        )

print("\nLogs saved to tool_logs_lab12.jsonl")
```
